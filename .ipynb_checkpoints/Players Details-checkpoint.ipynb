```python
%%sql
CREATE TABLE players (
    player_id INT PRIMARY KEY,
    player_name VARCHAR(50),
    country VARCHAR(30),
    team VARCHAR(50),
    role VARCHAR(30),
    batting_style VARCHAR(30),
    bowling_style VARCHAR(40),
    age INT,
    matches INT,
    runs INT,
    wickets INT,
    batting_avg DECIMAL(5,2),
    strike_rate DECIMAL(5,2),
    salary DECIMAL(8,2)
);
```

```python
%%sql
Select * from players;
```

```python
%%sql
Select Sum(runs) as Total_Runs from players;
```

```python
%%sql
Select Sum(runs) as Total_Runs from players where player_name = 'Suresh Raina';
```

```python
%%sql
Select player_name, runs from players where runs = (Select Max(runs)from players);
```

```python
%%sql
Select player_name, runs from players where country = 'India' order by runs desc limit 5;
```

```python
%%sql
Select * from players where country ='India' order by batting_avg desc limit 5; 
```

```python
%%sql
Select player_name , wickets from players order by wickets desc limit 5;
```

```python
%%sql
Select * from players order by runs limit 5;
```

```python
%%sql
Select player_name, runs from players where runs = (Select Min(runs) from players);
```

```python
%%sql
Select player_name,batting_avg from players where batting_avg >= 45;
```

```python
%%sql
Select country, Count(*) AS Total_Players from players where batting_avg >= 45 Group by country;
```

```python
%%sql
Select country, Sum(runs) as Total_Runs from players Group by country having Total_Runs > 30000;
```

```python
%%sql
Select country, Sum(wickets) as Total_Wickets from players Group by country having Total_Wickets > 1000;
```

```python
%%sql
Select country, Avg(strike_rate) as Avg_Strike_Rate from players Group by country having Avg_Strike_Rate > 130;
```

```python
%%sql
Select country, Min(age) from players Group by country having Min(age) < 28;
```

```python
%%sql
Select role,Count(*) as Total_Players from players Group by role;
```

```python
%%sql
Select role, Max(strike_rate) as Highest_Strike_Rate from players Group by role Order by Highest_Strike_Rate Desc;
```

```python
%%sql
Select player_name,Sum(runs) as Total_Runs from players Group by player_name Order by Total_Runs Desc Limit 10;
```

```python
%%sql
Select player_name,Sum(fifties) as Half_Centuries from players Group by player_name Order by Half_Centuries Desc Limit 10;
```

```python
%%sql
Select player_name, Sum(fours) as No_of_Fours from players Group by player_name Order by No_of_Fours Desc Limit 10;
```
