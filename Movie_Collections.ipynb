```python
%%sql
Create database MovieCollection;
```

```python
%%sql
use MovieCollection;
```

```python
%%sql
CREATE TABLE movie (
    id INT PRIMARY KEY,
    name VARCHAR(100),
    genre VARCHAR(50),
    budget_in_cr DECIMAL(10,2),
    collection_in_cr DECIMAL(10,2),
    rating DECIMAL(3,1),
    release_date DATE
);
```

```python
%%sql
CREATE TABLE actor (
    actor_id INT PRIMARY KEY,
    name VARCHAR(100),
    age INT
);
```

```python
%%sql
CREATE TABLE movie_cast (
    cast_id INT AUTO_INCREMENT PRIMARY KEY,
    movie_id INT,
    actor_id INT,
    role VARCHAR(30),

    FOREIGN KEY (movie_id) REFERENCES movie(id),
    FOREIGN KEY (actor_id) REFERENCES actor(actor_id)
);
```

```python
%%sql
Select * from movie;
```

```python
%%sql
Select * from actor;
```

```python
%%sql
Select * from movie_cast;
```

```python
%%sql
Select name, budget_in_cr, (budget_in_cr + 10) as new_budget from movie;
```

```python
%%sql
Select name, collection_in_cr, budget_in_cr, (collection_in_cr - budget_in_cr) as profit from movie;
```

```python
%%sql
Select * from  movie where (collection_in_cr - budget_in_cr) >= 50;
```

```python
%%sql
Select * from movie where genre in ('Action','Sports','Drama') and (collection_in_cr - budget_in_cr) >=30;
```

```python
%%sql
Select name,rating from movie where rating between 5 and 100;
```

```python
%%sql
Select name, Count(*) as Total_Movies from movie where genre = 'Action' and Year(release_date) = '2010' Group by name;
```

```python
%%sql
Select Count(*) as Total_Movies from movie where genre = 'Action' and Year(release_date) = '2016';
```

```python
%%sql
Select * from movie where Month(release_date) between 4 and 6;
```

```python
%%sql
Select monthname(release_date) as Month, Count(*) as Total_Movies from movie Group by monthname(release_date) Order by month;
```
