```python
%%sql
CREATE TABLE order_product (
    order_id INT,
    product_id INT,
    no_of_units INT NOT NULL,

    PRIMARY KEY (order_id, product_id),

    CONSTRAINT fk_order
        FOREIGN KEY (order_id)
        REFERENCES order_details(order_id)
        ON DELETE CASCADE
        ON UPDATE CASCADE,

    CONSTRAINT fk_product
        FOREIGN KEY (product_id)
        REFERENCES products(product_id)
        ON DELETE CASCADE
        ON UPDATE CASCADE
);
```

```python
%%sql
WITH CTE AS (
    SELECT *,
           LAG(people, 1) OVER (ORDER BY id) AS prev1,
           LAG(people, 2) OVER (ORDER BY id) AS prev2,
           LEAD(people, 1) OVER (ORDER BY id) AS next1,
           LEAD(people, 2) OVER (ORDER BY id) AS next2
    FROM Stadium
)
SELECT id, visit_date, people
FROM CTE
WHERE people >= 100
  AND (
        (prev1 >= 100 AND prev2 >= 100)
        OR
        (prev1 >= 100 AND next1 >= 100)
        OR
        (next1 >= 100 AND next2 >= 100)
      )
ORDER BY id;
```
