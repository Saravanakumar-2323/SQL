```python
%%sql
USE ElearningDB;
```

```python
%%sql
Create View Student_Course_View as 
Select Student.student_name, Course.course_name,Enrolls.course_id from Student
Inner join Enrolls on Student.student_id = Enrolls.student_id
Inner join Course on Enrolls.course_id = Course.course_id;
```

```python
%%sql
Select * from Student_Course_View;
```

```python
%%sql
Create view StudentNames as
Select student_name , student_id from Student;
```

```python
%%sql
Select * from StudentNames;
```
